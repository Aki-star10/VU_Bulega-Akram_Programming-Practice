```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("All libraries installed successfully!")

```

```python
# Create a small dictionary containing student information
data = {
    "Name": ["John", "Mary", "Alex", "Sarah", "David"],
    "Age": [20, 21, 19, 22, 20],
    "Score": [75, 88, 65, 92, 70]
}

# Convert the dictionary into a pandas DataFrame
df = pd.DataFrame(data)

# Display the data
df
```

```python
# Display the first 5 rows of the dataset
df.head()
```

```python
# Show the number of rows and columns
print("Shape of the data:", df.shape)

# Show the names of the columns
print("Columns:", df.columns.tolist())

# Show information about the data types
df.info()
```

```python
# Calculate the average score of all students
average_score = df["Score"].mean()

# Display the result
print("Average score:", average_score)
```

```python
# Find the highest score
highest_score = df["Score"].max()

# Find the lowest score
lowest_score = df["Score"].min()

print("Highest score:", highest_score)
print("Lowest score:", lowest_score)
```

```python
# Find students who scored more than 70
high_scores = df[df["Score"] > 70]

# Display the students
high_scores
```

```python
# Select students whose age is exactly 20
students_20 = df[df["Age"] == 20]

students_20
```

```python
# Create a new column called "Passed"
# A student passes if their score is 50 or higher
df["Passed"] = df["Score"] >= 50

# Display the updated DataFrame
df
```

```python
# Sort students from highest score to lowest score
sorted_df = df.sort_values(by="Score", ascending=False)

# Display the sorted data
sorted_df
```

```python
# Create a bar chart showing each student's score
plt.bar(df["Name"], df["Score"])

# Add a title to the graph
plt.title("Student Scores")

# Label the x-axis
plt.xlabel("Student")

# Label the y-axis
plt.ylabel("Score")

# Display the graph
plt.show()
```
